```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download (safe if already installed)
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def preprocess_text(text):
    text = clean_text(text)
    words = text.split()
    stop_words = set(stopwords.words('english'))
    words = [w for w in words if w not in stop_words]
    lemmatizer = WordNetLemmatizer()
    words = [lemmatizer.lemmatize(w) for w in words]
    return ' '.join(words)

texts = [
    "This product is amazing excellent perfect",
    "Worst product ever completely useless waste",
    "Good quality but very expensive and bad size"
]

for t in texts:
    cleaned = preprocess_text(t)
    vec = vectorizer.transform([cleaned])
    pred = model.predict(vec)
    prob = model.predict_proba(vec)

    print("\nTEXT:", t)
    print("Prediction:", pred)
    print("Prob:", prob)
```
